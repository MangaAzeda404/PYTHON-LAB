## 01 · Coleção de esmaltes
Some as três cores e descubra o tamanho da coleção de Luiza.
O operador + soma os valores. A atribuição (=) guarda o resultado em esmaltes_totais. print() exibe a mensagem; não calcula a soma por conta própria.


```python
esmaltes_azuis = 7
esmaltes_vermelhos = 18
esmaltes_brancos = 12

esmaltes_totais = esmaltes_azuis + esmaltes_vermelhos + esmaltes_brancos
print(f"Luiza tem {esmaltes_totais} esmaltes.")
```

## 02 · Pontuação de Pedro
Calcule os pontos da corrida e da natação no campeonato.
A corrida vale 3 pontos por quilômetro e a natação vale 5 pontos por volta. A multiplicação é avaliada antes da soma.


```python
corrida = 3
natacao = 5
quilometros = float(input("Quilômetros corridos: "))
voltas = int(input("Voltas na piscina: "))

total = corrida * quilometros + natacao * voltas
print(f"A pontuação total de Pedro é {total:g}")
```

## 03 · Jogos entre amigos
Distribua jogos em quantidades iguais e descubra a sobra.
Para quantidades não negativas, // indica quantos jogos inteiros cada amigo recebe e % calcula a sobra. A divisão é exata quando o resto é zero. O número de amigos deve ser maior que zero.


```python
total_jogos = int(input("Digite o total de jogos: "))
numero_amigos = int(input("Digite o número de amigos: "))

if total_jogos < 0 or numero_amigos <= 0:
    print("Informe jogos não negativos e pelo menos um amigo.")
else:
    jogos_por_amigo = total_jogos // numero_amigos
    resto = total_jogos % numero_amigos
    print(f"Cada amigo recebe {jogos_por_amigo} jogos.")
    print(f"Restam {resto} jogos.")
```

## 04 · Idade no futuro
Descubra a idade de uma pessoa depois de um número de anos.
input() recebe texto. int() converte uma entrada válida em inteiro. A soma é armazenada em idade_futura. Na f-string, as chaves inserem os valores na mensagem.


```python
idade_atual = int(input("Digite sua idade atual: "))
anos_adicionar = int(input("Quantos anos você quer adicionar? "))

idade_futura = idade_atual + anos_adicionar
print(f"Daqui a {anos_adicionar} anos, você terá {idade_futura} anos.")
```

## 05 · Área do triângulo
Use base e altura para calcular a área em unidades quadradas.
A fórmula é base × altura ÷ 2. É preciso converter as entradas de texto em números. float() permite medidas decimais; int() também funciona quando as medidas são inteiras. Use a mesma unidade na base e na altura.


```python
base = float(input("Digite a base do triângulo: "))
altura = float(input("Digite a altura do triângulo: "))

if base <= 0 or altura <= 0:
    print("A base e a altura devem ser positivas.")
else:
    area = base * altura / 2
    print(f"A área do triângulo é {area:g}.")
```

## 06 · Limite do carrinho
Verifique se dois itens respeitam o limite de 20 kg.
O operador <= inclui o limite: exatamente 20 kg também é permitido. if executa o primeiro bloco quando a condição é verdadeira; else executa o outro quando é falsa. Aqui float() também aceita pesos decimais.


```python
peso_item1 = float(input("Digite o peso do item 1: "))
peso_item2 = float(input("Digite o peso do item 2: "))

if peso_item1 + peso_item2 <= 20:
    print("Os itens cabem no carrinho.")
else:
    print("Os itens excedem o peso permitido.")
```

## 07 · Capacidade do cinema
Confira se um grupo pode entrar em um cinema de 100 lugares.
A ordem é: receber a quantidade, verificar a condição, exibir a mensagem do if ou do else. A indentação define quais instruções pertencem a cada bloco. O exemplo considera o cinema vazio e avalia somente o tamanho do grupo.


```python
total_pessoas = int(input("Quantas pessoas estão no grupo? "))

if total_pessoas <= 100:
    print("Todos podem entrar.")
else:
    print("O grupo excede a capacidade.")
```

## Operadores de comparação
`>` maior que; `<=` menor ou igual; `==` igual; `<` menor que. Para x = 5 e y = 8, os resultados são False, True, False e True.


```python
x = 5
y = 8
print("x > y:", x > y)
print("x <= y:", x <= y)
print("x == y:", x == y)
print("x < y:", x < y)
```
