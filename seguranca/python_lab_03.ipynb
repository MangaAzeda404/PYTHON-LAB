## Transformar uma senha
Cada dígito de “1111” é convertido em inteiro, recebe +2 e volta a ser texto. Os resultados são concatenados.

**Resposta: C) I, III e IV. Saída: 3333.**

I: a string pode ser percorrida dígito a dígito. II: falsa, o código concatena os resultados; não soma tudo para obter 12. III: int() permite a soma numérica. IV: cada 1 vira 3.

```python
base = "1111"
senha_final = ""
for digito in base:
    calculo = int(digito) + 2
    senha_final += str(calculo)
print(senha_final)
```

## Proteger o nome Bia
A posição de cada letra no alfabeto é usada para encontrar sua substituta no alfabeto secreto.

**Resposta: bia → hap**

b: índice 1 → h; i: índice 8 → a; a: índice 0 → p. ord(char) - ord('a') calcula o índice de 0 a 25. O trecho original pressupõe letras de a a z.

```python
alfabeto_secreto = "phqgiumeaylnofdxjkrcvstzwb"
nome = "bia"
nome_protegido = ""
for char in nome:
    posicao = ord(char) - ord('a')
    nome_protegido += alfabeto_secreto[posicao]
print(nome_protegido)
```

## Organizar a criptografia
Percorra a senha, teste a letra, calcule o índice e substitua. No caso contrário, preserve o caractere.

**Resposta: for → if → posicao → dicionario[posicao] → else → letra**

A indentação determina os blocos. else fica alinhado com if. Letras maiúsculas, números, espaços e símbolos são mantidos.

```python
dicionario = "phqgiumeaylnofdxjkrcvstzwb"
def codifica(senha):
    senha_criptografada = ""
    for letra in senha:
        if "a" <= letra <= "z":
            posicao = ord(letra) - ord("a")
            senha_criptografada += dicionario[posicao]
        else:
            senha_criptografada += letra
    return senha_criptografada

print(codifica("bia123!"))
```

## Organizar criar_hash
Lucas recebe um texto, inicia um acumulador em zero, soma o código de cada caractere e devolve o total após o laço.

**Resposta: def → valor = 0 → for → soma → return**

return fica fora do for para devolver a soma completa. ord() retorna o ponto de código Unicode; nas letras ASCII utilizadas nos exemplos, ele coincide com o valor ASCII.

```python
def criar_hash(texto):
    valor = 0
    for letra in texto:
        valor = valor + ord(letra)
    return valor

print(criar_hash("ROMA"))
```

## ROMA e AMOR: colisão
Bia soma os valores de ord(). ROMA e AMOR contêm as mesmas letras em ordem diferente.

**Resposta: D) As duas palavras geram 303: colisão.**

R=82, O=79, M=77, A=65; total 303. A: os códigos não são gastos. B: não há pesos por posição. C: reordenar letras não causa erro. D: entradas diferentes produzem a mesma saída.

```python
def calcula_hash(palavra):
    total = 0
    for letra in palavra:
        total = total + ord(letra)
    return total

print(calcula_hash("ROMA"))  # 303
print(calcula_hash("AMOR"))  # 303
```
